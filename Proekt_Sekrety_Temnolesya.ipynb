### 2.3. Наличие пропусков в данных



Абсоютное и относительное значение пропусков

```python
abs_value=df.isna().sum()
```

```python
otn_value=df.isna().sum()/len(df)
print(f"Абсолютное значение:{abs_value}\nОтносительное значение:{otn_value}")
```

Для продаж характерно отсутсвие значений, так как какие-то игры могли не продаваться в опредленных странах.
Для название игры отсутсвие значения не критично, скорее всего была ошибка в переносе данных.
Отсутсвующие оценки критиков и оценки пользователей можно вывести как среднее всех оценок.

```python
df['name']=df.dropna(subset=['name'])
```

```python
def sale_group(row):
    if pd.isna(row['eu_sales']):
        group = df[(df['platform'] == row['platform']) & 
                   (df['year_of_release'] == row['year_of_release'])]
        return group['eu_sales'].mean()
    else:
        return row['eu_sales']
    
df['eu_sales']=df.apply(['eu_sales'],axis=1)
```

```python
df.isna().sum()
```

## 3. Фильтрация данных

Коллеги хотят изучить историю продаж игр в начале XXI века, и их интересует период с 2000 по 2013 год включительно. Отберите данные по этому показателю. Сохраните новый срез данных в отдельном датафрейме, например `df_actual`.

## 4. Категоризация данных
    
Проведем категоризацию данных:


```python
df['user_score'] = pd.to_numeric(df_actual['user_score'], errors='coerce')
```

```python
df['user_score_category']=pd.cut(df_actual['user_score'],bins=[0,3,8,10],
                                 labels=["Низкая оценка","Средняя оценка","Высокая оценка"])
```

```python
print(df)
```

```python
df['critic_score'] = pd.to_numeric(df_actual['critic_score'], errors='coerce')
```

```python
df['critic_score_category']=pd.cut(df_actual['critic_score'],bins=[0,30,80,100],
                                   labels=["Низкая оценка","Средняя оценка","Высокая оценка"])
```

```python
print(df)
```

```python
grouped_user = df.groupby(['user_score_category'])['name'].count()
print(grouped_user)
```
