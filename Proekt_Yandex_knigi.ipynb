## 2. Проверка гипотезы в Python

Гипотеза звучит так: пользователи из Санкт-Петербурга проводят в среднем больше времени за чтением и прослушиванием книг в приложении, чем пользователи из Москвы. Попробуйте статистически это доказать, используя одностороннюю проверку гипотезы с двумя выборками:

- Нулевая гипотеза H₀: Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.

- Альтернативная гипотеза H₁: Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.

<div class="alert alert-block alert-success">✔️
    

__Комментарий от ревьюера №1__

Гипотеза корректна

```python
active_m=df[df['city']=='Москва']['hours'].reset_index()
active_s=df[df['city']=='Санкт-Петербург']['hours'].reset_index()
active_m.describe()
```

```python
active_s.describe()
```

Максимальное количество часов прослушивания составляет 978 часов, необходимо убрать выбросы. Оставим 90 процентов значений

<div class="alert alert-block alert-success">✔️
    

__Комментарий от ревьюера №1__

Да, обосновано

```python
quan_s=active_s['hours'].quantile(0.9)
active_s=active_s[active_s['hours']<=quan_s]
```

```python
quan_m=active_m['hours'].quantile(0.9)
active_m=active_m[active_m['hours']<=quan_m]
```

```python
active_s
```

```python
active_m
```

```python
alpha = 0.05 
metric_a=active_m['hours']
metric_b=active_s['hours']
stat_ttest, p_value_ttest = ttest_ind(
    metric_a,
    metric_b,
    alternative='less',
    equal_var=False
)

print(f'pvalue={p_value_ttest}')

if p_value_ttest>alpha: 
    print('Нулевая гипотеза находит подтверждение!')
    text_interpretation='в друг группах не различается'
else:
    print('Нулевая гипотеза не находит подтверждения!')
    text_interpretation = 'в Санкт-Петербурге больше'
print(f'Средняя активность пользователей {text_interpretation}')
```

Оцените достаточность выборки для получения статистически значимых результатов A/B-теста. Заданные параметры:

- базовый показатель конверсии — 30%,

- мощность теста — 80%,

- достоверность теста — 95%.

```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

alpha = 0.05  # Уровень значимости
beta = 0.2  # Ошибка второго рода, часто 1 - мощность
power = 1-beta  # Мощность теста
p1 = 0.3 # Базовый показатель конверсии
mde = 0.03  # Минимальный детектируемый эффект
p2=p1+mde

# Инициализируем класс NormalIndPower
power_analysis = NormalIndPower()
effect_size = (p2 - p1) / np.sqrt((p1*(1-p1) + p2*(1-p2))/2)
# Расчёт размера выборки
sample_size = power_analysis.solve_power(
    effect_size = effect_size,
    power = power,
    alpha = alpha,
    ratio = 1 # Равномерное распределение выборок
)

print(f"Необходимый размер выборки для каждой группы: {int(sample_size)}")
```

```python
from statsmodels.stats.proportion import proportions_ztest
n_a, n_b = all_user_a, all_user_b ## размеры выборок A и B
m_a = user_a_purchase

m_b = user_b_purchase # количество успехов (покупок) в группе B


    
alpha = 0.05 # На каком уровне значимости проверяем гипотезу о равенстве 

stat_ztest, p_value_ztest = proportions_ztest(
    [m_a, m_b],
    [n_a, n_b],
    alternative='smaller' # так как H_1: p_a < p_b
)
p_value_ztest


if p_value_ztest > alpha:
    print(f'pvalue={p_value_ztest} > {alpha}')
    print('Нулевая гипотеза находит подтверждение!')
else:
    print(f'pvalue={p_value_ztest} < {alpha}')
    print('Нулевая гипотеза не находит подтверждения!')
```
